```python
import csv
import random
from datetime import datetime, timedelta
import json
```

```python
origin = ['local', 'imported']
organic_or_inorganic = ['organic', 'inorganic']
#fruits, vegetables, dairy, meat, fish, bread
fruits = [ "Apple", "Banana", "Cherry", "Date", "Grape", "Kiwi", "Lemon", "Mango", 
          "Orange", "Papaya", "Peach", "Pear", "Pineapple", "Plum", "Pomegranate", 
          "Raspberry", "Strawberry", "Watermelon", "Blueberry", "Blackberry", "Dragonfruit", 
          "Cantaloupe", "Apricot", "Fig", "Tangerine", "Lychee" ]

meats = [ "Chicken", "Beef", "Pork", "Lamb", "Turkey", "Duck", "Goat", "Venison", 
         "Rabbit", "Bison", "Veal", "Mutton", "Ham", "Sausage", "Salami", "Bacon" ]

bread_types = [
    "White", "Whole Wheat", "Multigrain", "Rye", "Sourdough",
    "Baguette", "Ciabatta", "Focaccia", "Brioche", "Bagel",
    "Pita", "Naan", "Tortilla", "Flatbread", "Hoagie Roll",
    "Sub Roll", "Kaiser Roll", "Ciabatta Roll", "Texas Toast", "English Muffin"
]

dairy_products = [
    "Milk", "Heavy Cream",
    "Cheddar", "Mozzarella", "Swiss", "Provolone",
    "Parmesan", "Feta", "Goat", "Blue", "Brie",
    "Yogurt", "Greek Yogurt", "Sour Cream", "Ricotta",
    "Butter", "Cream Cheese", "Whipped Cream", "Ice Cream",
    "Eggs"
]

seafood = [
    "Salmon", "Tuna", "Cod", "Haddock", "Halibut",
    "Trout", "Snapper", "Bass", "Sardine",
    "Anchovy", "Swordfish", "Grouper", "Pollock",
    "Lobster", "Crab", "Clam", "Oyster", "Mussel",
    "Scallop", "Squid", "Octopus", "Shrimp", "Prawn"
]

vegetables = [
    "Carrot", "Broccoli", "Cauliflower", "Spinach", "Lettuce",
    "Kale", "Arugula", "Cabbage", "Zucchini", "Cucumber",
    "Tomato", "Bell Pepper", "Jalapeño", "Onion", "Garlic",
    "Potato", "Sweet Potato", "Beet", "Radish", "Turnip",
    "Celery", "Asparagus", "Leek", "Green Bean", "Pea",
    "Mushroom", "Eggplant", "Brussels Sprout", "Artichoke"
]

def data_gen(produce_type, produce_list, id):
    current_id = id
    current_origin = random.choice(origin)
    current_org = random.choice(organic_or_inorganic)
    current_type = produce_type
    current_item = random.choice(produce_list)
    currrent_price = round(random.uniform(0.01, 100), 2)

    # data = [current_id,
    #         current_item,
    #         current_type,
    #         current_origin,
    #         currrent_price,
    #         current_org]
    
    data = [current_id,
            current_item,
            current_type,
            currrent_price,]
    
    return data
```

```python
data_rows = []

for i in range(0, 167):
    data_rows.append(data_gen("fruit", fruits, i+1))

for i in range(167, 334):
    data_rows.append(data_gen("vegetable", vegetables, i+1))

for i in range(334, 501):
    data_rows.append(data_gen("dairy", dairy_products, i+1))

for i in range(501, 668):
    data_rows.append(data_gen("meat", meats, i+1))

for i in range(668, 835):
    data_rows.append(data_gen("seafood", seafood, i+1))

for i in range(835, 1002):
    data_rows.append(data_gen("bread", bread_types, i+1))
```

```python
with open("training_data.csv", "w", newline="") as file:
    writer = csv.writer(file)
    # writer.writerow(
    #     ["id", "item", "type", "origin", "price", "organic_or_inorganic"]
    # )
    writer.writerow(
        ["id", "item", "type", "price"]
    )
    writer.writerows(data_rows)
```
